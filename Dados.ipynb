```python
# dicionario com python
# 1 - todo dicionario é um data frame(df)

turmaad6 = {
    # chave é unitario
    # valor por não ser unitario
    "professores" : ["Felipe", "Luyz","Priscila"],
    "aluno": ["Gabriel","Alessandra","Stefani"]
}

print(turmaad6["professores"][1])
```

```python
# vamos criar as funções
# ver os dados
# adicionar novo professor
# adicionar novo aluno
# remover aluno
# remover professor

def ver_dados():
  print("--- professores cadastrados ---")
  for i in turmaad6["professores"]:
    print(i)
  print("--- Alunos Cadastrados---")
  for i in turmaad6["aluno"]:
    print(i)

def add_prof():
  novo_professor = input("Informe o nome do Professor:")
  # chama a lista .append(dado)
  turmaad6["professores"].append(novo_professor)
  print(f"O novo professor {novo_professor} foi adicionado com sucesso!")
  ver_dados()

def add_aluno():
  novo_aluno = input("Informe o nome do aluno:")
  turmaad6["aluno"].append(novo_aluno)
  print(f"O aluno{novo_aluno} foi adicionado com sucesso")
  ver_dados()

def remover_aluno():
  posicao = 1
  for i in turmaad6['aluno']:
    print(f"{posicao} - {i}")
    posicao +=1
  remover = int(input("informe a posição do aluno"))
  # para remover
  turmaad6["aluno"].pop(remover -1)
  ver_dados()

def remover_professor():
  posicao = 1
  for i in turmaad6["professores"]:
    print(f"{posicao} - {i}")
    posicao +=1

    remover = int(input("Informe a posição do professor"))
    turmaad6["professores"].pop(remover -1)
    ver_dados()
```

```python
remover_professor()
```
